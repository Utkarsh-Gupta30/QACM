<h3>Implement a multi-qubit circuit with CNOT and SWAP gates </h3>

```python
from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector
qc = QuantumCircuit(3)
# qc.h(0)
# qc.h(1)
# qc.h(2)
qc.swap(0, 1)
qc.cx(1, 2)
display(qc.draw("mpl"))
state = Statevector.from_instruction(qc)
display(state.draw('latex'))
```
