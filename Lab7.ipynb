<h3>Implement and run Grover’s algorithm to search for a marked state</h3>

```python
from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator
qc = QuantumCircuit(2,2)
qc.h([0,1])
# Oracle
qc.x(1)
qc.cz(0,1)
qc.x(1)
# Diffusion
qc.h([0,1])
qc.x([0,1])
qc.cz(0,1)
qc.x([0,1])
qc.h([0,1])
# Measurement
qc.measure([0,1], [0,1])
display(qc.draw('mpl'))
# Simulator
sim = AerSimulator()
result = sim.run(qc, shots = 1000).result()
count = result.get_counts()
print(count)
```
